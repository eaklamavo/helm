## The `helm install` Command

The basic syntax is:
```
helm install <RELEASE-NAME> <CHART> [flags]
```

- `<RELEASE-NAME>` — the name *you* give to this running instance (e.g., `my-nginx`)
- `<CHART>` — where to get the chart from (repo, OCI URI, local path, or `.tgz`)

> **Reference:** [Helm Quickstart Guide](https://helm.sh/docs/intro/quickstart/) — https://helm.sh/docs/intro/quickstart/

---

## Option A: Install from an OCI Registry (Helm 3.8+)

````python
# Install the 'podinfo' chart directly from an OCI registry
# No 'helm repo add' needed — OCI registries work with the oci:// prefix
helm install my-podinfo oci://ghcr.io/stefanprodan/charts/podinfo \
  --version 6.11.2 \
  --namespace default \
  --create-namespace
````

## Option B: Install from a Classic Repository

````python
# Make sure the bitnami repo is added (from Lesson 1.2)
helm repo add bitnami https://charts.bitnami.com/bitnami 2>/dev/null || true
helm repo update

# Install nginx from Bitnami
helm install my-nginx bitnami/nginx \
  --namespace helm-study \
  --create-namespace \
  --set replicaCount=1
````

---

## Inspect the Release

After installing, Helm gives you a summary. You can always look it up again:

````python
# List all releases (across all namespaces)
helm list --all-namespaces
````

````python
# Get detailed status of a specific release
helm status my-nginx --namespace helm-study
````

````python
# See the exact Kubernetes YAML that was deployed for this release
helm get manifest my-nginx --namespace helm-study
````

````python
# See the values used for this release (default + any overrides you set)
helm get values my-nginx --namespace helm-study
````

---

## Preview Before You Install: `--dry-run`

Before applying real changes, use `--dry-run` to render the templates and inspect them without touching the cluster:

````python
# Dry run: render templates and show what would be deployed — nothing is sent to Kubernetes
helm install preview-nginx bitnami/nginx \
  --namespace helm-study \
  --dry-run \
  --set replicaCount=2 2>/dev/null | head -60
````

> **Gotcha:** `--dry-run` still contacts the Kubernetes API server to validate the templates. Use `helm template` (covered in Lesson 2.3) for fully offline rendering.

## Upgrading a Release

When you want to change configuration or update the chart version, use `helm upgrade`:

````python
# Upgrade my-nginx to use 2 replicas
helm upgrade my-nginx bitnami/nginx \
  --namespace helm-study \
  --set replicaCount=2

# Check the revision history — you should see REVISION 2 now
helm history my-nginx --namespace helm-study
````

## Rolling Back

Changed your mind? Roll back to a previous revision:

````python
# Roll back to revision 1 (the original install)
helm rollback my-nginx 1 --namespace helm-study

# Check history again — rollback creates a new revision
helm history my-nginx --namespace helm-study
````

---

## Uninstalling a Release

````python
# Uninstall removes all Kubernetes resources AND the release history
helm uninstall my-nginx --namespace helm-study

# Verify it is gone
helm list --namespace helm-study
````

````python
# TIP: Use --keep-history to remove resources but preserve history for auditing
# helm uninstall my-nginx --namespace helm-study --keep-history
# After --keep-history, the status is 'uninstalled' and you can helm rollback to restore it
echo "Uninstall with --keep-history preserves rollback capability"
````

## Summary

| Command | What It Does |
|---|---|
| `helm install <name> <chart>` | Install a chart as a new release |
| `helm upgrade <name> <chart>` | Upgrade an existing release |
| `helm rollback <name> <revision>` | Roll back to a previous revision |
| `helm uninstall <name>` | Remove all resources and release history |
| `helm list` | List installed releases |
| `helm status <name>` | Get release status |
| `helm history <name>` | Show revision history |
| `helm get manifest <name>` | Show Kubernetes YAML that was applied |
| `--dry-run` | Preview rendered templates without applying |
