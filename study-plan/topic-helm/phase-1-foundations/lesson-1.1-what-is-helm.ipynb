## The Problem Helm Solves

Imagine you want to deploy a web application to Kubernetes. Without Helm you need to:

1. Write a `Deployment` YAML
2. Write a `Service` YAML
3. Write a `ConfigMap` YAML
4. Write an `Ingress` YAML
5. Apply them in the right order: `kubectl apply -f ...`
6. Do it all over again for dev, staging, and production — with slightly different values each time

That gets painful fast. **Helm** solves this by letting you bundle all those YAML files into a single, versioned, reusable package called a **chart**.

Think of Helm like `apt` (Ubuntu) or `brew` (macOS) — but for Kubernetes applications.

---

## Key Terms

| Term | Definition |
|---|---|
| **Chart** | A package of Kubernetes resource templates + default configuration. Like a Docker image, but for K8s apps. |
| **Release** | A running instance of a chart in your cluster. Installing the same chart twice creates two releases. |
| **Repository** | A place where charts are stored and shared (like Docker Hub, but for charts). |
| **Values** | Configuration you provide to customise a chart at install time (e.g., replica count, image tag). |
| **Template** | A YAML file with Go-template placeholders that Helm fills in using your values. |

---

## Helm's Place in the Kubernetes Workflow

```
Developer writes Chart
        |
        v
Chart published to Repository (e.g., Artifact Hub, OCI registry)
        |
        v
Operator runs: helm install my-app oci://registry/my-chart --values prod.yaml
        |
        v
Helm renders templates  -->  Kubernetes API  -->  Running Release
```

## Helm Versions

- **Helm 2** — required a server-side component called *Tiller*. Now deprecated.
- **Helm 3** — client-only, no Tiller. Releases are stored as Kubernetes Secrets. Much simpler.
- **Helm 4** (current as of 2026) — further improvements; most Helm 3 charts are still compatible.

> **Reference:** [Helm Official Docs](https://helm.sh/docs) — https://helm.sh/docs

## What a Chart Looks Like

Here is the directory structure of a minimal Helm chart:

````python
# This cell prints the typical structure of a Helm chart directory.
# You do not need Helm installed to run this — it just shows text.
echo "
mychart/
  Chart.yaml        # Metadata: name, version, description
  values.yaml       # Default configuration values
  charts/           # Sub-charts (dependencies) live here
  templates/        # Go-template YAML files
    deployment.yaml
    service.yaml
    _helpers.tpl    # Reusable template snippets
    NOTES.txt       # Post-install message shown to the user
"
````

## A Minimal Chart.yaml

The `Chart.yaml` file is the identity card of your chart:

````python
# Write a minimal Chart.yaml to disk so you can inspect it.
cat > /tmp/Chart.yaml << 'EOF'
apiVersion: v2          # Use v2 for Helm 3+
name: mychart           # Chart name
description: A minimal Helm chart example
type: application       # 'application' (default) or 'library'
version: 0.1.0          # Chart version (SemVer)
appVersion: "1.0.0"     # Version of the application inside the chart
EOF

cat /tmp/Chart.yaml
````

### Key Fields Explained

- `apiVersion: v2` — Required for Helm 3+. Tells Helm which chart format to expect.
- `version` — The version of the *chart itself* (follows [SemVer](https://semver.org/)). Increment this when you change the chart.
- `appVersion` — The version of the *application* the chart deploys (e.g., your Docker image tag). Informational only.

> **Important:** Always quote `appVersion` values like `"1.0.0"` — unquoted values like `1.0` are treated as floats by YAML parsers and may cause errors.

## The Release Lifecycle

When you run `helm install`, Helm does this:

1. Reads the chart
2. Merges your values with the chart's defaults
3. Renders the templates into plain Kubernetes YAML
4. Sends the YAML to the Kubernetes API
5. Stores release metadata as a Kubernetes Secret (so it can upgrade/rollback later)

Helm tracks every release revision. You can always roll back to a previous state.

## Summary

- **Helm** is the Kubernetes package manager — it packages, versions, installs, and manages K8s apps.
- A **chart** is the package; a **release** is a running installation of a chart.
- Helm 3 is client-only (no Tiller), stores state in K8s Secrets, and is the version in active use.
- `Chart.yaml` is the identity card; `values.yaml` holds defaults; `templates/` holds the K8s YAML with Go-template placeholders.
