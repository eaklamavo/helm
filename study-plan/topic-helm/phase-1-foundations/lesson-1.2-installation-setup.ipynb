## Step 4: How Helm Knows Which Cluster to Use

Helm uses the **same kubeconfig** file as `kubectl` — typically `~/.kube/config`. It reads the *current context* to know which cluster to talk to.

You can override this with:
- `--kube-context <name>` — use a specific context
- `--kubeconfig /path/to/config` — use a different kubeconfig file
- `--namespace <ns>` or `-n <ns>` — target a specific namespace

````python
# Show all available contexts in your kubeconfig
kubectl config get-contexts
````

## Step 5: Helm Environment Variables

Helm uses a few environment variables you should know:

````python
# Show Helm's environment configuration
helm env
````

Key variables:

| Variable | Purpose |
|---|---|
| `HELM_DATA_HOME` | Where Helm stores plugins and starters |
| `HELM_CACHE_HOME` | Where downloaded chart archives are cached |
| `HELM_CONFIG_HOME` | Where `repositories.yaml` lives |
| `KUBECONFIG` | Path to kubeconfig (inherited from kubectl) |

---

## Step 6: Add Your First Chart Repository

A **chart repository** is a collection of charts hosted on an HTTP server or OCI registry. The classic way to add one is with `helm repo add`.

> **Note:** With Helm 3+ and OCI registries, repositories are becoming less common for large publishers. But `helm repo` is still widely used for traditional repos.

````python
# Add the Bitnami chart repository (a popular community source)
helm repo add bitnami https://charts.bitnami.com/bitnami

# Update local cache of chart metadata from all added repos
helm repo update

# List all configured repositories
helm repo list
````
